### what is  Encoding 🤔

**When you’ve got categorical data (like colors, gender, city names, etc.), machines can’t handle words directly — they only understand numbers.So we have to encode those categories into numerical form.**


### Nominal/Label Encoding
**Nominal encoding is the process of assigning a unique numerical value to each category in a categorical feature.**
| Color |                                   
| :---: |
|  Red  |
|  Blue |
| Green |
|  Red  |
| Green |


| Color | Encoded_Value |
| :---: | :-----------: |
|  Red  |       1       |
|  Blue |       2       |
| Green |       3       |
|  Red  |       1       |
| Green |       3       |


### One Hot Encoding
**One-hot encoding converts each category into a binary vector, creating separate columns for every unique category.**
| Color | Red | Blue | Green |
| :---: | :-: | :--: | :---: |
|  Red  |  1  |   0  |   0   |
|  Blue |  0  |   1  |   0   |
| Green |  0  |   0  |   1   |
|  Red  |  1  |   0  |   0   |
| Green |  0  |   0  |   1   |




### Sparse Matrix
**A sparse matrix is a large matrix where most elements are zeros, so instead of storing every value, only the non-zero elements and their positions are kept. This saves a ton of memory and computation time, especially when dealing with data that naturally has many zeros — like one-hot encoded features, text data (bag of words or TF-IDF), or graph connections. In contrast, a dense matrix stores every element, which is only efficient when most values are non-zero Sparse matrices make big data processing practical and fast.**


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
# Create a sample DataFrame
data = {'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red']}
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
## Create an instance of OneHotEncoder
encoder = OneHotEncoder()

```

```python
# Perform fit and transform
encoded=encoder.fit_transform(df[['Color']]).toarray()
```

```python
import pandas as pd
encoder_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['Color']))
encoder_df
```

```python
encoder.transform([['Red']]).toarray()
```

```python
pd.concat([df, encoder_df], axis = 1)
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

lbl_encoder = LabelEncoder()
lbl_encoder.fit_transform(df[['Color']])

```

```python

lbl_encoder.transform([['Red'], ['Green'], ['Blue']])
```

### Ordinal Encoding
**Ordinal encoding is a data preprocessing technique used to convert categorical variables (text labels) into numerical values, but unlike regular label encoding, it’s specifically designed for ordered categories — categories that have a meaningful ranking or hierarchy.**

### Example
| Satisfaction     | Encoded |
| ---------------- | ------- |
| Very Unsatisfied | 1       |
| Unsatisfied      | 2       |
| Neutral          | 3       |
| Satisfied        | 4       |
| Very Satisfied   | 5       |

**Now, the algorithm knows that 5 > 4 > 3 > 2 > 1, and that the difference between “Very Unsatisfied” and “Satisfied” is meaningful in the context of order, even if not strictly linear.**


```python
from sklearn.preprocessing import OrdinalEncoder
df = pd.DataFrame({
    'Satisfaction' : ['Very Unsatisfied', 'Unsatisfied', 'Neutral', 'Satisfied', 'Very Satisfied']
})
```

```python
df
```

```python
## Create an instance of OneHotEncoder
encoder = OrdinalEncoder(categories=[['Very Unsatisfied', 'Unsatisfied', 'Neutral', 'Satisfied', 'Very Satisfied']])
# Perform fit and transform
encoded = encoder.fit_transform(df[['Satisfaction']])
encoded
```

```python
encoder.transform([['Very Satisfied']])
```
