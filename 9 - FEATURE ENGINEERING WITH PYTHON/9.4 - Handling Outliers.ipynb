### What is box plot ?
**A Box Plot is a data visualization that summarizes a dataset’s distribution.**

**Box: The box spans from the first quartile (Q1) to the third quartile (Q3) representing the interquartile range (IQR). This box contains the middle 50% of the data with a line inside it representing the median (Q2).**

**Whiskers: The lines extending from the box called whiskers shows the range of the data. They extend from Q1 to the minimum value and from Q3 to the maximum value excluding outliers.**

**Outliers: Data points that fall outside the whiskers are considered outliers. These are marked separately and represent values that are significantly higher or lower than the rest of the data.**



### Creating a Box Plot in Python

```python
import pandas as pd
import matplotlib.pyplot as plt

data = [0, 1, 2, 3, 4, 5, 6]
df = pd.DataFrame(data, columns=['Num'])
df
```

```python
plt.figure(figsize = (10, 7)) # The first number (10) → width of the figure (horizontal size).
                              # The second number (7) → height of the figure (vertical size).

df.boxplot()
```

```python
import numpy as np

data = [0, 1, 2, 3, 4, 5, 6]

Minimum, Q1, Median, Q3, Maximum = np.quantile(data, [0, 0.25, 0.5, 0.75, 1.0])

print("Minimum:", Minimum)
print("Q1:", Q1)
print("Median:", Median)
print("Q3:", Q3)
print("Maximum:", Maximum)
IQR = Q1 - Q3 
print("IQR:", IQR )
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data= [0, 1, 2, 3, 4, 5, 10]
df= pd.DataFrame(data, columns=['Num'])

plt.figure(figsize=(10, 7))
df.boxplot()
plt.show()

### Here the max value is 5 because the third quartile is 4.5 and the interquartile range is (4.5-1.5)=3. 
# So, 1.5*3 is 4.5 and third quartile(4.5)+4.5=9. So 10 is larger than the limit 9 thus it becomes an outlier.
```

```python
import numpy as np
data= [0, 1, 2, 3, 4, 5, 10,12,20,30]

Minimum, Q1, Median, Q3, Maximum = np.quantile(data, [0, 0.25, 0.5, 0.75, 1.0])
IQR = Q3 - Q1
lower_fence = Q1 - 1.5*(IQR)      ## Lower Bound
higher_fence = Q3 + 1.5*(IQR)     ## Upper Bound
print("Minimum:", Minimum)
print("Q1:", Q1)
print("Median:", Median)
print("Q3:", Q3)
print("Maximum:", Maximum)
print("IQR:", IQR )
print("Lower Fence",lower_fence)
print("Hiher Fence",higher_fence)
```
