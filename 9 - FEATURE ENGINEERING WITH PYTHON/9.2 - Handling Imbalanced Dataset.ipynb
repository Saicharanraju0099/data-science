```python
import numpy as np
import pandas as pd
np.random.seed(123)
n_sample = 1000
classification_0_ratio = 0.9
n_classification_0 = int(n_sample * classification_0_ratio)
n_classification_1 = n_sample - n_classification_0
```

```python
n_classification_0, n_classification_1
```

```python
# Creating a dataframe with imbalanced dataset
classification_0 = pd.DataFrame({
    'feature1' : np.random.normal(loc=0, scale=1, size=n_classification_0),
    'feature2' : np.random.normal(loc=0, scale=1, size=n_classification_0),
    'target'   : [0] * n_classification_0
    
})
classification_1 = pd.DataFrame({
    'feature1' : np.random.normal(loc=2, scale=1, size=n_classification_1),
    'feature2' : np.random.normal(loc=2, scale=1, size=n_classification_1),
    'target'   : [1] * n_classification_1
})
```

```python
df=pd.concat([classification_0, classification_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
## upsampling the minority class
df_minority = df[df['target'] ==1]
df_majority = df[df['target'] ==0]
```

```python
from sklearn.utils import resample
df_minority_upsampled=resample(df_minority,replace=True,n_samples=len(df_majority),random_state=243)
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled=pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```

### Down sampling

```python
np.random.seed(123)
n_sample = 1000
classification_0_ratio = 0.9
n_classification_0 = int(n_sample * classification_0_ratio)
n_classification_1 = n_sample - n_classification_0

classification_0 = pd.DataFrame({
    'feature1' : np.random.normal(loc=0, scale=1, size=n_classification_0),
    'feature2' : np.random.normal(loc=0, scale=1, size=n_classification_0),
    'target'   : [0] * n_classification_0
    
})
classification_1 = pd.DataFrame({
    'feature1' : np.random.normal(loc=2, scale=1, size=n_classification_1),
    'feature2' : np.random.normal(loc=2, scale=1, size=n_classification_1),
    'target'   : [1] * n_classification_1
})

df=pd.concat([classification_0, classification_1]).reset_index(drop=True)
```

```python
# Separate majority and minority classes
df_majority = df[df.target == 0]
df_minority = df[df.target == 1]
```

```python
df_majority_downsampled = df_majority.sample(
    n=len(df_minority),       # same number as minority
    random_state=123
)
```

```python
df_majority_downsampled.shape
```

```python
df_majority_downsampled.head()
```

```python
df_downsampled = pd.concat([df_majority_downsampled, df_minority])
```

```python
print(df_downsampled['target'].value_counts())
```
