### Univariate Analysis
**"Uni" = one variable.**

**Goal: understand the distribution, spread, and outliers of a single column**

**Univariate = "How does this one column behave?"**

**Think of it as reading the personality of each variable**

```python
## Pairplot To See The Relationship Between Different Features In The Dataset
sns.pairplot(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

num_cols = df.select_dtypes(include='number').columns

plt.figure(figsize=(15, 10))
for i, col in enumerate(num_cols, 1):
    plt.subplot(len(num_cols)//3 + 1, 3, i)
    sns.histplot(df[col].dropna(), kde=True, color='skyblue')
    plt.title(col)
    plt.tight_layout()
plt.show()
```

```python
## Univariate For Categorical Columns in The Dataset 
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Identify categorical columns
cat_cols = df.select_dtypes(exclude='number').columns

if len(cat_cols) == 0:
    print("No categorical columns found.")
else:
    plt.figure(figsize=(15, 10))

    for i, col in enumerate(cat_cols, 1):
        plt.subplot(len(cat_cols)//3 + 1, 3, i)
        sns.countplot(data=df, y=col, order=df[col].value_counts().index, palette='pastel')
        plt.title(col)
        plt.tight_layout()
    
    plt.show()

```

```python
### Univariate Analysis Categorical Plot 
## there simply weren’t any categorical columns to plot.
sns.countplot(x='quality', data=df, palette='pastel')
plt.title('Wine Quality Distribution')
plt.show()
```

### Bivariate Analysis

**"Bi" = two variables.**

**Goal: study relationships — how one variable changes with another.**

```python
## analyze relationships between features and quality

cols_to_plot = ['alcohol', 'volatile acidity', 'sulphates', 'citric acid']

plt.figure(figsize=(14, 10))
for i, col in enumerate(cols_to_plot, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(x='quality', y=col, data=df, palette='magma')
    plt.title(f'{col} vs Quality')
    plt.tight_layout()
plt.show()
```

### Multivariate Analysis
**To see how multiple features interact at once**


```python
## Multivariate Analysis
## To see how multiple features interact at once
## matrix of scatterplots showing pairwise relationships between every numerical column in your dataframe
## pairplot = one-stop visual correlation map — perfect for spotting clusters, trends, and outliers across all features in one glance.
sns.pairplot(df, hue='quality', diag_kind='kde', palette='husl')
plt.show()
```

### OUTPUT EXPLANATION
**Seaborn automatically assigned colors from the 'husl' palette (a rainbow-like gradient)**


**Typically it goes like this:**

**Red or dark pink dots → Quality 3 or 4 (lowest)**

**Orange / yellow → Quality 5 or 6 (medium)**

**Green / teal / blue → Quality 7 or 8 (higher)**

**The warm colors (reds/oranges) cluster in regions representing poorer-quality wines.**

**The cooler colors (greens/blues) appear where the wines have better quality.**

**This is a Seaborn pairplot that shows how all the numerical features in the wine quality dataset relate to each other and to wine quality"**

**Each little square (scatterplot) compares two features — for example, alcohol vs. density, or volatile acidity vs. citric acid.**

**Each point is a wine sample, and its color represents the quality level (from 3 to 8).**

**The diagonal plots show the distribution of each single feature and and the off-diagonal plots show relationships between features.**

**Wines with higher alcohol (bottom-right corner) usually have higher quality — the colors for quality 7 and 8 cluster at higher alcohol values.**

**Wines with higher volatile acidity (top part of the grid) usually have lower quality — those dots skew toward the lower-quality colors.**

**Density and alcohol move in opposite directions — as alcohol increases, density decreases.**

**Other features like citric acid and sulphates show weaker, but still visible, positive trends with quality.**

**This is a pairplot showing relationships between all the chemical properties of the wines in the dataset.**

**Each dot represents one wine sample, and the color of the dot indicates the wine’s quality score — lower-quality wines are on one end of the color spectrum and higher-quality wines on the other.**
