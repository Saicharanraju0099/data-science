### Date

```python
## Feature Engineering and Data Preprocessing
# (This section would include handling missing values, encoding categorical variables, feature scaling, etc.)
# parse original 'date' (format DD-MM-YYYY) and extract day, month, year


# Step 2: Split and safely convert to int
df['Date'] = pd.to_numeric(df['date'].str.split('-').str[0], errors='coerce').astype('Int64')
df['Month'] = pd.to_numeric(df['date'].str.split('-').str[1], errors='coerce').astype('Int64')
df['Year'] = pd.to_numeric(df['date'].str.split('-').str[2], errors='coerce').astype('Int64')


```

```python
df.head()
```

```python
df.info()
```

```python
## Drop Date Column because we have extracted Day, Month, Year
df.drop(columns=['date'], inplace=True)
## df.drop('Date', axis=1, inplace=True)    # If you want to drop the 'Date' column instead
```

### Arrival Time

```python
df['Arrival_Hour'] = df['arr_time'].str.split(':').str[0].astype(int)
df['Arrival_Minute'] = df['arr_time'].str.split(':').str[1].astype(int)
```

```python
## Drop the original arrival time column because we have extracted hour and minute
df.drop(columns=['arr_time'], inplace=True)
```

### Depature Time

```python
df['Departure_Hour'] = df['dep_time'].str.split(':').str[0].astype(int)
df['Departure_Minute'] = df['dep_time'].str.split(':').str[1].astype(int)
```

```python
## Drop the original Departure time column because we have extracted hour and minute
df.drop(columns=['dep_time'], inplace=True)
```

### Time Taken 

```python
df['hours'] = (df['time_taken'].str.split(' ').str[0].str.replace('h', '', regex=False).astype(str))
df['minutes'] = (df['time_taken'].str.split(' ').str[-1].str.replace('m', '', regex=False).astype(str))
```

### One Hot Encoder Data Preprocessing Technique

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
```

```python
encoder.fit_transform(df[['airline','from','to']])   ## Sparse Matrix Output
```

```python
encoder.fit_transform(df[['airline','from','to']]).toarray()
```

```python
encoded_df = pd.DataFrame(
    encoder.transform(df[['airline', 'from', 'to']]).toarray(),
    columns=encoder.get_feature_names_out(['airline', 'from', 'to'])
)
```

```python
# Reset index so both have same index alignment
encoded_df.index = df.index
```

```python
# Drop original categorical columns
df = df.drop(['airline', 'from', 'to'], axis=1)
```

```python
# Merge encoded columns with the original DataFrame
X_final = pd.concat(
    [df.drop(columns=['airline', 'from', 'to', 'price'], errors='ignore'), encoded_df],
    axis=1
)
```
