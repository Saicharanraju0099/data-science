### Data Cleaning

```python
df['Reviews'].unique()        ## It is object type but should be int type
```

```python
df['Reviews'].str.isnumeric().sum()
```

```python
df[~df['Reviews'].str.isnumeric()]     ## getting the rows which are non numeric there is only one row
```

```python
df_copy = df.copy()        ## creating a copy of the original dataset
```

```python
df_copy=df_copy.drop(df_copy.index[10472])
```

```python
df_copy[~df['Reviews'].str.isnumeric()]     ## checking again after dropping the row
```

```python
## Convert the Reviews column to int type
df_copy['Reviews']=df_copy['Reviews'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy['Size'].unique()      ## checking the unique values in Size column
```

```python

# Clean up non-numeric patterns
df_copy['Size'] = df_copy['Size'].replace('Varies with device', np.nan)  # Replace text with NaN
df_copy['Size'] = df_copy['Size'].str.replace(',', '', regex=True)       # Remove commas
df_copy['Size'] = df_copy['Size'].str.replace('M', '000', regex=True)    # Replace M with 000
df_copy['Size'] = df_copy['Size'].str.replace('k', '', regex=True)       # Remove k

# Now safely convert
df_copy['Size'] = pd.to_numeric(df_copy['Size'], errors='coerce')
```

```python
df_copy['Size']
```

```python
df_copy.info()
```

```python
df_copy['Installs'].unique()      ## checking unique values in Installs column
```

```python
df_copy['Price'].unique()     ## checking unique values in Price column
```

```python
char_to_remove = ['+', ',', '$']
cols_to_clean = ['Installs', 'Price']
for  item in char_to_remove:
    for col in cols_to_clean:
        df_copy[col] = df_copy[col].str.replace(item, '')
```

```python
df_copy['Price'].unique() 
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Installs'] = df_copy['Installs'].astype('int')
df_copy['Price'] = df_copy['Price'].astype('float')
```

```python
df_copy.info()
```

```python
## Handling The last Updated Column
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.info()                ## final info of the cleaned dataset
```

```python
##  Exporting the cleaned dataset
df_copy.to_csv("/Users/saicharan/Datascience/11 - Explority Data Analysis/Data/PlayStore Dataset/PlayStore_Cleaned.csv", index=False) ## exporting without the index
```

### Explore data

```python
numeric_features = [feature for feature in df_copy.columns if df_copy[feature].dtype != 'O']
categorical_features = [feature for feature in df_copy.columns if df_copy[feature].dtype == 'O']

## print Features 
print('we have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nwe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

```python
## proportion of categorical and numerical features
for col in categorical_features:
    print(df_copy[col].value_counts(normalize=True)*100)
    print("-"*50)
```

```python
# proportion of numerical features
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
plt.suptitle('Numerical Feature Distributions', fontsize=16, fontweight='bold', alpha = 0.8, y=1.02)
for i in range(len(numeric_features)):
    plt.subplot(3,3,i+1)
    sns.kdeplot(df_copy[numeric_features[i]],color='red',shade=True)
    plt.xlabel(numeric_features[i])
    plt.tight_layout()
```

### observation
**Rating, lastupdated and year is left skewed while others are right skewed**

```python
## categorical features observation

for i in range(0,len(categorical_features)):
    plt.figure(figsize=(12,6))
    plt.suptitle('Categorical Feature Distributions', fontsize=16, fontweight='bold', alpha = 0.8, y=1.02)
    category = ['Type','Content Rating']
    for i, col in enumerate(category):
     plt.subplot(1, len(category), i+1)
     sns.countplot(y=df_copy[col], color='green', palette='Set2')
     plt.xlabel(col)
     plt.xticks(rotation=0)
     plt.tight_layout() 
    
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

category = ['Type', 'Content Rating']
plt.figure(figsize=(12, 16))  # Adjust height for vertical stacking
plt.suptitle('Categorical Feature Distributions', fontsize=16, fontweight='bold', alpha=0.8, y=1.02)

for i, col in enumerate(category):
    plt.subplot(len(category), 1, i+1)
    sns.countplot(x=df_copy[col], palette='Set2')
    plt.xlabel(col)
    plt.xticks(rotation=0)

plt.tight_layout(rect=[0, 0, 1, 0.97])  # Ensures the title is visible
plt.show()
```
