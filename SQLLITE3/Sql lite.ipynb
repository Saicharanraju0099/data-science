```python
import sqlite3
```

```python
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
cursor.execute('''
Create Table If Not Exists Employee(
      id Integer Primary Key,
      name Text Not Null,
      age Integer,
      department Text,
      salary Integer
               )
               
''')

# commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from Employee
               ''')
```

```python
# insert the data in to table
cursor.execute('''
Insert Into Employee (id, name, age, department, salary)
               Values(1, 'sai',20,'IT', 50000)
               
               ''')

cursor.execute('''
Insert Into Employee (id, name, age, department, salary)
               Values(2, 'charan',20,'CSE', 60000)
               
               ''')

cursor.execute('''
Insert Into Employee (id, name, age, department, salary)
               Values(3, 'raju',22,'Data scientist', 100000)
               
               ''')

# commit the changes
connection.commit()
```

```python
# Query the data from the table
cursor.execute('Select * from Employee')
rows = cursor.fetchall()

# print the results

for row in rows:
    print(row)
```

```python
# update the data in the table
cursor.execute('''
UPDATE Employee
set age = 21
where name = 'sai'
               
''')
# commit the changes
connection.commit()
```

```python
# Query the data from the table
cursor.execute('Select * from Employee')
rows = cursor.fetchall()

# print the results

for row in rows:
    print(row)
```

```python
# delete the data from the table
cursor.execute('''
DELETE from Employee
where name = 'charan'
               ''')

# commit the changes
connection.commit()
```

```python
# Query the data from the table
cursor.execute('Select * from Employee')
rows = cursor.fetchall()

# print the results

for row in rows:
    print(row)
```
