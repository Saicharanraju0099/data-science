### Pandas Dataframe and series
**Series:A Pandas Series is a one-dimensional labeled array that can hold data of any type (integer, string, float, etc.) with an associated index.**

**Data Frame: A Pandas DataFrame is a two-dimensional, tabular data structure with labeled rows and columns, where each column is essentially a Series.**

```python
import pandas as pd
```

```python
# Series
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series of Numbers \n",series)
print(type(series))
```

```python
# Create a series with dictionary
data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)
print(type(series_dict))
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index=index)
```

### Definition: Data Manipulation and Analysis with Python

**Data manipulation and analysis with Python means using Python libraries (like pandas, NumPy, etc.) to clean, transform, organize, and study data so we can extract useful information, find patterns, or gain insights.**

**Data Manipulation = adjusting the raw data → cleaning, filtering, sorting, adding new columns, deleting bad data, handling missing values**

**Data Analysis = interpreting that cleaned data → calculating stats, finding trends, summarizing, visualizing, making decisions**

```python
# Filling missing values with mean of column
# Filling Missing Values with means of column
data_set['Unit_Price'] = data_set['Unit Price'].fillna(data_set['Unit Price'].mean())
data_set
```

```python
# Renaming Columns
Renamed_data=data_set.rename(columns={'Date':'Oreder_Date'})
Renamed_data.head(5)
```

```python
# Change Datatypes
data_set['Units Sold_New'] = data_set['Units Sold'].astype('float')
data_set.head(5)
```

```python
data_set['Units Sold_New'] = data_set['Units Sold'].astype('int')
data_set.head(5)
```

```python
# Change Datatypes and apply function
data_set['Units Sold_New'] = data_set['Units Sold'].apply(lambda x: x * 2)
data_set.head(5)
```

```python
# Data Aggregating and Grouping
data_set.head()
```

```python
Grouped_data = data_set.groupby('Product Name')['Unit Price'].mean().reset_index()
print(Grouped_data)
```

```python
grouped_sum=data_set.groupby(['Product Category','Region'])['Unit Price'].sum()
print(grouped_sum)
```

```python
data_set.groupby(['Product Category','Region'])['Unit Price'].mean()
```

```python
## Aggregate Multiple Functions
Grouped_agg=data_set.groupby('Region')['Unit Price'].agg(['sum','mean','count'])
Grouped_agg
```

```python
# Merging And Joining DataFrames
df1 = pd.DataFrame({'key':['A','B','C'],'Value1':[1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'],'Value1':[4,5,6]})


```

```python
df1
```

```python
df2
```

```python
# Merging DataFrames on key Columns
pd.merge(df1,df2, on='key', how='inner')
```

```python
pd.merge(df1,df2, on='key', how='outer')
```

```python
pd.merge(df1,df2, on='key', how='left')
```

```python
pd.merge(df1,df2, on='key', how='right')
```
