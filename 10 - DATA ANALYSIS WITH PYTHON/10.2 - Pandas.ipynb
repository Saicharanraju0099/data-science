### Pandas Dataframe and series
**Pandas:A Pandas Series is a one-dimensional labeled array that can hold data of any type (integer, string, float, etc.) with an associated index.**

**Data Frame: A Pandas DataFrame is a two-dimensional, tabular data structure with labeled rows and columns, where each column is essentially a Series.**

```python
import pandas as pd
```

```python
# Series
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series of Numbers \n",series)
print(type(series))
```

```python
# Create a series with dictionary
data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)
print(type(series_dict))
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index=index)
```

```python
# DataFrame
# Create a dataframe from dictionary of list
data = {
    'Name':['Krish','bharath','ismail'],
    'Age': [20,24,28],
    'city':["bangalore",'Amaravati','Telangana']
}
df = pd.DataFrame(data)
print(df)
print(type(df))

```

```python
# create a data frame from list of dictionaries
data = [
    {'Name':'Krish','Age':20,'city':'vizag'},
    {'Name':'bharath','Age':23,'city':'Amaravati'},
    {'Name':'Krishna','Age':28,'city':'Guntur'},
]
df = pd.DataFrame(data)
print(df)
print(type(df))
```
