```python
import seaborn as sns
```

```python
# basic plotting with seaborn
tips = sns.load_dataset("tips")
tips
```

```python
## creating a scatter plot
import matplotlib.pyplot as plt 
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title("scatter plot of total bill vs tip")
plt.show()
```

```python
# line plot
sns.lineplot(data=tips, x='size', y='total_bill')
plt.title("line plot of total bill vs tip")
plt.show()
```

```python
#categort plots
# bar plot
sns.barplot(data=tips, x='day', y='total_bill')
plt.title("the total bill on different days")
plt.show()
```

```python
# box plot
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title("the total bill on different days")
plt.show()
```

```python
# violin plot
sns.violinplot(data=tips, x='day', y='total_bill')
plt.title("The total bill on different days")
plt.show()
```

```python
# histogram 
sns.histplot(tips['total_bill'], bins=10, kde=True)
plt.title("Histogram of total bill")
plt.show()
```

```python
# kde plot
sns.kdeplot(tips['total_bill'], shade=True,fill=True)
```

```python
# pair plot 
sns.pairplot(tips)   #this will tell the relationship between all the numerical columns
plt.show()
```

```python
# heat map
sns.heatmap(tips[['total_bill','tip','size']].corr())

```

```python
# heat map
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
# heat map
sns.heatmap(corr, annot=True, cmap='coolwarm')
```
