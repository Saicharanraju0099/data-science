### Adding and removing files using lambda and costum functions

```python
bios_data['Height Category'] = bios_data['height_cm'].apply(lambda x: 'Short' if x < 165 else('Average' if x < 185 else 'Tall'))
```

```python
bios_data.head()
```

```python
def categorize_athlete(row):
    if row['height_cm'] < 175 and row['weight_kg'] < 70:
        return 'Short and Light'
    elif row['height_cm'] < 185 and row['weight_kg'] <= 80:
        return 'Average'
    else:
        return 'Large and Heavy'
bios_data['Category'] = bios_data.apply(categorize_athlete, axis=1)
```

```python
bios_data.head()
```
