### Matplotlib
*Matplotlib is a powerful and widely-used open-source plotting library for the Python programming language. Its primary purpose is to create static, animated, and interactive visualizations of data.*

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]
# Line plot
plt.plot(x,y,color = 'black', linestyle = '--', marker = 'o', markerfacecolor = 'red', markersize = 10)
plt.xlabel('Numbers')
plt.ylabel('Squares')
plt.title('Basic Line Plot')
plt.grid(True)
```

```python


x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1,2,3,4,5]
plt.figure(figsize = (10,5))
plt.subplot(1,2,1)
plt.plot(x,y1,color = 'blue', marker = 'o')
plt.title('Plot 1: Squares')
```

```python
# Bar plot
categories = ['A', 'B', 'C', 'D', 'E']
values = [3,7,3,9,10]
plt.bar(categories, values, color = 'green')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Basic Bar Plot')
```

### Histogram

**Matplotlib is a popular Python library for creating static, interactive, and animated visualizations, and it provides a straightforward way to generate histograms.**

```python
import numpy as np
data = np.random.randn(1000)
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Distribution of Sample Data')
plt.show()
```

```python
data = [2,4,5,6,2,3,4,5,7,8,2,3,5]
plt.hist(data, bins=5, color='skyblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Distribution of Sample Data')
plt.show()
```

```python
# Scatter plot
x = [1,2,3,4,5,6,7]
y = [2,3,4,5,6,7,8]
plt.scatter(x,y, color = 'purple', marker='^')
plt.xlabel('X - Axis')
plt.ylabel('Y - Axis')
plt.title('Basic Scatter Plot')
plt.show()
```

```python
labels = ['Agriculture', 'Mining', 'Power Distribution', 'Education']
sizes = [15, 30, 45, 10]
explode = (0,0,0.1,0)
colors = ['gold', 'lightcoral', 'lightskyblue', 'lightgreen']
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=colors,explode=explode, shadow=True)
plt.title('Distribution of Categories')
plt.axis('equal')
plt.show()
```

```python
x = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
y = [10,15,7,10,30]
plt.plot(x,y, color = 'Black', marker = 'o',label = '2025 Coffee Sales')
plt.xlabel('Days of the week')
plt.ylabel('Sales per Day')
plt.title('Bakery Sales This Week')
plt.legend(loc='upper left', fontsize = 12)
plt.grid(color = 'grey', linestyle = ':', linewidth = 1.2)
plt.ylim(0,40)
plt.xticks(['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'])
plt.show()
```

### Subplot

```python
x = [1,2,3,4,5]
y = [15,20,25,25,30]
plt.subplot(1,2,1)   # 1 row, 2 columns, 1rd plot
plt.plot(x,y)
plt.title('Line Plot')
plt.subplot(1,2,2)  # 1 row, 2 columns, 2nd plot
plt.bar(x,y,color = 'cyan')
plt.tight_layout()
plt.title('Bar Plot')
plt.show()

```

```python
x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.plot(x,y,color = 'magenta', marker = 'o')
plt.title('Sample Line Plot')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.savefig('line_plot.png',dpi = 300,bbox_inches = 'tight')   # Save the plot as a PNG file with 300 dpi resolution in Same folder location
plt.show()
```
