**The map() function in Python is a built-in function used to apply a given function to each item in an iterable (or multiple iterables) and return an iterator that yields the results.**

**Syntax:**
**map(function, iterable(s))**

**function:
The function to be applied to each item of the iterable(s). This can be a built-in function, a user-defined function, or a lambda function.**

**iterable(s):**
**One or more iterable objects (e.g., lists, tuples, sets, strings) whose elements will be passed as arguments to the function. If multiple iterables are provided, the function must accept a corresponding number of arguments, and map() will process elements from each iterable in parallel until one of the iterables is exhausted.**

**Return Value:**

**The map() function returns a map object, which is an iterator. To view the results as a list or another collection type, the map object must be explicitly converted (e.g., using list(), tuple()).**

```python
def square(n):
    return n ** 2
square(9)
```

```python
numbers = [1,2,3,4,5,6]
list(map(square,numbers))
```

```python
#lambda function with map
numbers = [1,2,3,4,5,6]
list(map(lambda x:x**2,numbers))
```

```python
# map multiple iterables
numbers1=[1,2,3]
numbers2=[4,5,6]
sum_of_numbers = list(map(lambda x,y:x+y,numbers1,numbers2))
print(sum_of_numbers)
```

```python
# map() to convert list of strings in to integers
str_numbers = ['1','2','3','4','5']
int_numbers = list(map(int,str_numbers))
print(int_numbers)
```

```python
Programming_languages = ["c","c++","java","javascript","python"]
capitalize = list(map(str.upper,Programming_languages))
print(capitalize)
```

```python
def get_name(person):
    return person['name']
people = [
    {"name":"saicharan","age":20,"origin":"INDIA"},
    {"name":"sai","age":23,"origin":"INDIA"}
]
list(map(get_name,people))
```
