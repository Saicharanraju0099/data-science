**Conditional statements in Python are used to execute certain blocks of code based on specific conditions.**

**1.if Statement
Definition: The if statement evaluates a condition (an expression that returns True or False). If the condition is True, the indented block of code under it is executed.**

```python
a = 10
if a > 5:
    print("a is greater  than 5")
```

```python
age = 20
if age >= 18:
    print("You are eligible to vote in the election.")
```

```python
age = 16
if age > 18:
    print("You are eligible to vote in the election.")
else:
    print("you are a minor")
```

**2.elif Statement
Definition: The elif (short for "else if") statement is used to check additional conditions if the previous if or elif conditions are False. It must follow an if statement.**

```python
# elif statements
age = 20
if age > 13:
    print("You are a teenager.")
elif age > 18:
    print("You are an adult.")
else:
    print("You are a child.")
```

**3.The 'if-elif-else' Statement**
**'elif' (short for “else if”) allows checking multiple conditions sequentially. The 'else' at the end is optional and runs if none of the conditions are true.**

```python
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: D")
```

**4.Nested Conditional Statements
An if statement inside another if or else block is called a nested if. Used for checking multiple levels of conditions.**

```python
num = int(input("Enter a number: "))
if num >= 0:
    if num == 0:
        print(num,"The number is zero.")
    if num % 2 == 0:
        print( num,"The number is even.")
    else:
        print(num,"The number is odd.")
else:
    print(num,"The number is negitive.")
```

**Practical Examples**

```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

```python
age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age.")
elif age <= 5:
    print("Ticket Price: Free")
elif age <= 12:
    print("Ticket Price: ₹100")
elif age < 60:
    print("Ticket Price: ₹200")
else:
    print("Ticket Price: ₹150")
```
