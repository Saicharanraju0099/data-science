### File Operations in python

**File operations in Python involve interacting with files on a computer's file system, allowing for reading, writing, and managing data. The fundamental steps for file operations are: opening the file, performing read/write operations, and closing the file.**


```python
## Reading a whole File
with open('example.txt','r') as file:
    content = file.read()
    print(content)
  
```

```python
## Read a file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) # This Strip function removes the new line character
```

```python
# writing a file(overwriting)
with open('example.txt','w') as file:
    file.write("Happy coding \n")
    file.write("Python is known language to me")
    
```

```python
# write a file(Without Overwriting)
with open('example.txt','a') as file:
    file.write("\n Append Operation Taking place")
```

```python
# writing a list of lines to a file
lines = ['First line \n','Second line \n','Third line \n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
# binary Files
some_bytes = b'\x00\x01\x02\x03\x04'

with open("example.bin", "wb") as binary_file:
  
    binary_file.write(some_bytes)
```

```python
with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
## Read a content from source text file write it to a destination file
with open('example.txt','r') as file:
    content = file.read()
with open('destination.txt','w') as file:
    file.write(content)
```

```python
# Read a text file and count number of lines,words,characters
file_path = "example.txt" 

line_count = 0
word_count = 0
char_count = 0

with open(file_path, 'r', encoding='utf-8') as file:
    for line in file:
        line_count += 1
        words = line.split()
        word_count += len(words)
        char_count += len(line)  

print(f"Lines: {line_count}")
print(f"Words: {word_count}")
print(f"Characters: {char_count}")
```

```python
with open('example.txt', 'w+') as file1:
    file1.write("Hello world\n")
    file1.write("This is a new line\n")
    
    # move the file cursor to the beginning
    file1.seek(0)
    
    # read while file is still open
    content = file1.read()

print(content)
```
