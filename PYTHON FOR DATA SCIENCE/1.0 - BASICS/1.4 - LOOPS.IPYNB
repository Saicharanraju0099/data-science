**🌀1. The for Loop — Iterating Over Sequences
Python’s for loop behaves like a for-each loop, iterating directly over elements of an iterable (like lists, strings, tuples) or numeric ranges.**

🧩 Syntax:
 for variable in sequence:
    # code block


```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10, 0, -2):
    print(i)
```

```python
n = 4
for i in range(0, n):
    print(i)
```

```python
str = "Coding"
for char in str:
    print(char)
```

```python
li = "Python","Java","C++"
for lang in range(len(li)):
    print(li[lang])
```

**While Loop in Python
In Python, a while loop is used to execute a block of statements repeatedly until a given condition is satisfied. When the condition becomes false, the line immediately after the loop in the program is executed.**

🧩 Syntax:
while condition:


    # code block




```python
count = 0
while count < 5:
    print("Count is:", count)
    count += 1
```

while condition:

     # execute these statements


else:

     # execute these statements

```python
count = 0
while count < 5:
    print("Count is:", count)
    count += 1
else:
    print("Count is reached 5, exiting loop.")
```
