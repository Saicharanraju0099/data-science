**✅ 1. What is a Variable?
A variable is a name that refers to a value stored in memory. In Python, variables are created the moment you assign a value to them.**

✅ Rules for Naming Variables

1.Must start with a letter or underscore (_)

2.Can contain letters, digits, and underscores

3.Case-sensitive: myVar, MyVar, and MYVAR are different.

4.Cannot use Python reserved keywords (if, for, class, etc.)

```python
# declaring and assigning variables in python
age = 20
height = 6.3
name = "jordhan"
is_athlete = True
# printing the variables
print(age)
print(height)
print(name)
```

```python
# variable names case sensitive
first_name = "john"
last_name = "jackson"
```

```python
# Invalid variable names
# 1age = 20
# first-name = "john"
# @name = "jordhan"
```

```python
# case sensitive 
name = "sai"
Name = "charan"
name == Name 
```
