Operators in Python are special symbols or keywords that perform operations on one or more operands (values or variables). They are fundamental for manipulating data, performing calculations, making comparisons, and controlling program flow.

```python
# Arithmetic operators in python
a = 10
b = 5
result = a + b, a - b, a * b, a / b, a % b, a ** b, a // b
print("result  is : ", result)
```

```python
# comparision operators in python
a = 10
b = 10
a == b
```

```python
str1 = "hello"
str2 = "hello"
str1 == str2
```

```python
str1 != str2
```

```python
# Greater than operator
a = 10
b = 100
a > b
```

```python
a < b
```

```python
# greater than or equal to operator
num1 = 10
num2 = 20
num1 >= num2
```

**Logical operators**

```python
#and,not,or operators
 
x = True
y = False
result = x and y
print(result)
```

```python
x = True
y = True
result = x and y
print(result)
```

```python
# or operator
x = True
y = False
result = x or y
print(result)
```

```python
x = False
y = False
result = x or y
print(result)
```

```python
# Not operator
x = True
result = not x
print(result)
```

```python
x = False
result = not x
print(result)
```
