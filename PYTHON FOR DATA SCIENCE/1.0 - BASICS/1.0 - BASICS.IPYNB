Python syntax refers to the set of rules that dictate how Python code must be written and structured for the interpreter to understand and execute it correctly

```python
# case sensitivity : python is case-sensitive
name = "sai"
Name = "charan"
print(name)
print(Name)
```

**Indentation:
Unlike many other programming languages that use braces or semicolons to define code blocks, Python uses indentation (whitespace at the beginning of a line) to indicate the structure of code blocks (e.g., within if statements, for loops, or function definitions). Consistent indentation is crucial, as incorrect indentation will lead to IndentationError**

```python
age = 25
if age > 20:
    print(age)
print("This is outside of if block")
```

**Comments**

```python
# This is single line comment
print("hello world")
```

```python
# Line continuation:
# using backslash to continue a statement on th next line
sum = 1 + 2 + 3 + 4 \
      + 5 + 6 + 7
print(sum)
```

```python
# multiple statements on a single line
x = 10; y = 20; z = x + y
print(z)
```

```python
# understanding semantics in python
# variable assignment
age = 30
name = "Mohan"

print(type(age))
print(type(name))
```

```python
# Type inference 
var = 10
print(type(var))
var = "hello"
print(type(var))
```

```python
# Indentation
if True:
    print("This is Correct Indentation")
    if False:
        print("This will not print")
    print("This will print")
print("This is outside of if block")
```
