**A Python lambda function is a small, anonymous function defined using the lambda keyword. Unlike a standard function defined with def, a lambda function does not have a name and is restricted to a single expression, which is implicitly returned.**

**Syntax :**
**lambda arguments : expression**

**Arguments: These are the input values or parameters that the lambda function accepts. Similar to arguments in a regular def function, they are variables that hold the values passed into the function for processing. A lambda function can take any number of arguments, including zero, one, or multiple arguments, separated by commas.**

**Expression: This is the single, executable piece of code that the lambda function evaluates and returns. Unlike regular functions that can contain multiple statements and complex logic, a lambda function is restricted to a single expression. The result of this expression is implicitly returned by the lambda function.**

**lambda arguments : expression**

**Here, lambda is the keyword defining the anonymous function, arguments are the inputs, and expression is the single operation performed and returned.**

```python
# Normal Function
def add(x,y):
    return x + y
# calling a function
add(3,6)
```

```python
lambda x,y:x+y   
```

```python
addition = lambda x,y:x+y
print(addition(1,2))
```

```python
def even(num):
    if num % 2 == 0:
        print("even")
    else:
        print("odd")
even(23)
```

```python
even_number = lambda x:x%2 == 0
even_number(100)
```

```python
def addition(x,y,z):
    return x + y + z
addition(1,2,3)
```

```python
sum = lambda x,y,z:x+y+z
sum(1,2,3)
```

```python
name = "Krishna"
name_1 = lambda func:func.upper()
print(name_1(name))
```

```python
n = lambda x: "Positive Number" if x > 0 else "Negitive" if x < 0 else "zero"
print(n(5))   
print(n(-3))  
print(n(0))
```

```python
# normal Function
def sqdef(x):
    return x ** 2
print(sqdef(3))
```

```python
# Using lambda
square = lambda x: x ** 2
print(square(3))
```

**Lambda with List Comprehension**

**Syntax**

**[expression for item in iterable if condition]**

```python
# using a for loop 
a = [1,2,3,4,5]
result = []
for val in a:
    result.append(val ** 2)
print(result)
```

```python
# Using a lamba function
x = [1,2,3,4,5]
squaring_list = [val ** 2 for val in x]
print(squaring_list)
```

```python
li = [lambda args=x:args*10 for x in range(1,5)]
for i in li:
    print(i())
```
