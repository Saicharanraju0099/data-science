```python
## Function with multiple parameters
def add(a,b):
    """ This function is used for Sum of numbers """
    return a + b
# Calling a function
add(2,4)
```

```python
# Default parameters 
def greet(name):
    print(f"Hello {name} Welcome to ITC hotel")

greet("saicharan")
```

```python
# Default parameters 
def greet(name="saicharan"):
    print(f"Hello {name} Welcome to ITC hotel")

greet()
```

```python
# Variable Length Arguments
# positional and keyword arguments
def print_numbers(*sai):  # This parameter could be anything
    for numbers in sai:
        print(numbers)
print_numbers(1,2,3,4,5,6,7,"Coding")
```

```python
def print_numbers(*args):  # This parameter could be anything
    for numbers in args:
        print(numbers)
print_numbers(1,2,3,4,5,6,7,"Coding")
```

```python
# keywords arguments
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
# calling a function
print_details(Name="sai",age="20",country="India")
```

```python
# keyword and positional Arguments
def print_details(*args,**kwargs):
    for numbers in args:
        print(f"Positional Arguments:{numbers}")

    for key,value in kwargs.items():
        print(f"Keyword Argument {key}:{value}")
# calling a function
print_details(1,2,3,"code", Name="sai",age="20",country="India")
```
