**Example 1** **Temperature Conversion**

```python
def temperature_converter(temp,unit):
    """This function converts the temperature between """
    if unit == 'C':
        return temp * 9/5 + 32 #celsius to fahrenheit
    
    elif unit == 'F':
        return temp -32 * 5/9 # fahrenheit to celsius
    else:
        return None
print(f"Celsius temperature to fahrenheit: ",temperature_converter(23,'C'))
print(f"Fahrenheit temperature to celsius: ",temperature_converter(40,'F'))


```

**Example 2: Password Strength Checker**

```python
def is_password_stronger(password):
    """This function checks password is strong or not"""
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '$@!$%*&+_'for char in password):
        return False
    else:
        return True
    
# calling the function
print(is_password_stronger("PassWd"))
print(is_password_stronger("Pass123@*"))

    
    
```

**Example 3 : Calculate the total cost of items in a shopping cart**

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    return total_cost



# Cart data
cart = [
    {'name':'surfexcel','price':180,'quantity':2},
    {'name':'freedom oil','price':170,'quantity':2},
    {'name':'oranges','price':80,'quantity':2}
]

# calling the function
total_cost = calculate_total_cost(cart)
print(total_cost)
```
