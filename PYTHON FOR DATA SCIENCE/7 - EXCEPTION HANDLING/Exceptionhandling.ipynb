### What is Exception Handling in Python?


**It’s basically Python’s way of saying:**
**Something went wrong… but instead of throwing your whole program in the trash, I’ll give you a chance to deal with it like an adult.**

**Instead of crashing on an error (like dividing by zero or opening a missing file), you can catch the error and handle it gracefully.**

### what are exceptions
**Exceptions in Python are basically runtime errors — signals that something unexpected happened while your program was running, and Python doesn’t know how to continue unless you handle it.**

**They occur during execution (runtime), not during compilation (Python doesn’t have a strict compile step anyway).**

### Examples:

**Trying to divide by zero → ZeroDivisionError**

**Accessing a missing file → FileNotFoundError**

**Converting "abc" to int → ValueError**

**Accessing an index out of range → IndexError**
