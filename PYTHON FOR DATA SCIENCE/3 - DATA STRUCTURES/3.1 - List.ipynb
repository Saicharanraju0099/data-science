**Python List**

**In Python, a list is a built-in dynamic sized array (automatically grows and shrinks). We can store all types of items (including another list) in a list**

```python
lst = []
print(type(lst))
```

```python
# Creating a Python list with different data types
a = [10, 20, "code", 40, True]
print(a)
```

```python
mixed_list = [10,20,"Hello",3.14,True]
print(mixed_list)
```

```python
## Accessing List elements
fruits = ["apple", "banana", "mango", "orange"]
print(fruits[0]) # Accessing the first element
print(fruits[1])
```

```python
print(fruits[1:])
print(fruits[1:3])
```

```python
# modifying list elements
fruits[1] = "kiwi"
print(fruits)
```

```python
# list methods 
fruits.append("Grapes") # Adding an element to the end of the list
print(fruits)
```

```python
fruits.insert(2, "Pine apple") # Inserting an element at a specific position
print(fruits)
```

```python
fruits.remove("apple") # Removing an element from the list
print(fruits)
```
