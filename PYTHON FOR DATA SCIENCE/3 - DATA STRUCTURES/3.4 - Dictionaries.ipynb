In Python, a dictionary is a built-in data structure that stores data in key-value pairs. It is a mutable, unordered (in Python versions before 3.7, ordered from 3.7 onwards), and indexed collection where each key must be unique and immutable

```python
empty_dict = dict()
empty_dict
```

```python
student = {"Name":"charan","Age":30,"Grade":"A"}
print(student)
```

```python
student = {"Name":"sai","Age":30,"Name":"tej"}
print(student)
```

```python
# accessing the dictionary elements
student = {"Name":"sai","Age":30,"Name":"tej","Grade":"A"}
print(student)
```

```python
print(student['Age'])
 
#Accessing using the get() method
print(student.get('Grade'))
print(student.get('Lastname'))
print(student.get('Firstname',"Not Available"))
```
