In Python, a dictionary is a built-in data structure that stores data in key-value pairs. It is a mutable, unordered (in Python versions before 3.7, ordered from 3.7 onwards), and indexed collection where each key must be unique and immutable

```python
empty_dict = dict()
empty_dict
```

```python
student = {"Name":"charan","Age":30,"Grade":"A"}
print(student)
```

```python
student = {"Name":"sai","Age":30,"Name":"tej"}
print(student)
```

```python
# accessing the dictionary elements
student = {"Name":"sai","Age":30,"Name":"tej","Grade":"A"}
print(student)
```

```python
print(student['Age'])
 
#Accessing using the get() method
print(student.get('Grade'))
print(student.get('Lastname'))
print(student.get('Firstname',"Not Available"))
```

```python
# modifying dictionaries 
# dict are mutable, you can add,update,delete key always must be unique
print(student)
```

```python
student["Age"] = 25
print(student)
student["address"] = "INDIA"
print(student)
```

```python
del student["Age"]
print(student)
```

```python
# Dict methods
keys = student.keys()
print(keys)
values = student.values()
print(values)

item = student.items()
print(item)
```

```python
# shallow copy 
student_copy=student
print(student)
print(student_copy)
```

```python
student["Name"] = "Charan"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## shallow copy
print(student_copy1)
print(student)
```

```python
student["Name"] = "bharath"
print(student_copy1)
print(student)
```
