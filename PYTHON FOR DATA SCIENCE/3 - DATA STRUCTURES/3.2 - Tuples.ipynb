In Python, a tuple is an ordered and immutable collection of items. This means that once a tuple is created, its elements cannot be changed, added, or removed. Tuples are defined by enclosing comma-separated elements within parentheses ().

```python
#Tuple creation
empty_tuple = ()
print(empty_tuple)
print(type(tuple))
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
numbers = tuple([1,2,3,4,5,6])
print(numbers)
```

```python
list([1,2,3,4,5,6])
```

```python
mixed_tuple = (1,2,3,"Hello","world",3.14,True)
print(mixed_tuple)
```

```python
# Immutable Nature of Tuples
# tuples are immutable, meaning they cannot be changed once assigned
lst = [1,2,3,4,5]
print(lst)
```

```python
lst[1] = "code"
print(lst)
```

```python
#Tuples are Im mutable
numbers[1] = "coding"
print(numbers) 
```

```python
# Tuple Methods
print(numbers)
print(numbers.count(1))
print(numbers.index(4))
```

```python
# packing and unpacking 
packed_tuple = "Hello",3.14,1,2,3
print(packed_tuple)
```

```python
# unpacking
a,b,c,d,e = packed_tuple
print(a)
print(b)
print(c)

```

```python
# unpacking with * 
numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
# Nested Tuple
nested_tuple = ([1,2,3],[4,5,6],[1,"Hello",3.14,"Aerospace"])
nested_tuple[0][0:3]
```

```python
nested_tuple = ([1,2,3],[4,5,6],[1,"Hello",3.14,"Aerospace"])
nested_tuple[2][-1:4]
```

```python
#iterating over nested tuples
for subtuple in nested_tuple:
    for item in subtuple:
        print(item,end = " ")
    print()
```
