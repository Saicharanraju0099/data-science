**Sets in Python are an unordered collection data type that stores unique elements. They are mutable, meaning elements can be added or removed after creation, but the individual elements within the set must be immutable (e.g., numbers, strings, tuples). Sets do not allow duplicate values; if an attempt is made to add a duplicate, it will be ignored.**

```python
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
emptyset = set()
print(emptyset)
print(type(emptyset))

```

```python
empty_set = set([1,2,3,4,5,6,7])
print(empty_set)
print(type(empty_set))
```

```python
set = {1,2,3,5,6,7,8,9}
```

```python
#Basic set operations 
# add and remove elements
set.add(9)
print(set)
set.add(8)
print(set)
```

```python
#remove the elements
set = ([1,2,3,4,5,6,7,1,2])
set.remove(2)
print(set)
set.remove(3)
print(set)
```

```python
my_set.discard(9)
print(my_set)
```

```python
# pop method
removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
## set membership
set_1 = {1,2,3,4,5}
print(3 in set_1)
print(6 in set_1)
```

```python
# mathematical operations
set1 = {1,2,3,4,5,6,7,8}
set2 = {9,10,11,12,5,6}

# set union
union_set = set1.union(set2)
print(union_set)
```

```python

#intersection set
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6,7}
set2 = {6,7,8,9,10}
print(set1.difference(set2))
```

```python
print(set1.difference_update(set2))
```

```python
print(set2.difference(set1))
```

```python
# symetric difference
set1.symmetric_difference(set2)
```

```python
set1 = {1,2,3}
set2 = {4,5,6}
print(set1.issubset(set2))
print(set1.issuperset(set2))
```

**Python Frozen Sets**

**Frozen sets in Python are immutable objects that only support methods and operators that produce a result without affecting the frozen set or sets to which they are applied. It can be done with frozenset() method in Python.**

```python
s = set(["a", "b","c"])

print("Normal Set")
print(s)
```

```python
# A frozen set
fs = frozenset(["e", "f", "g"])

print("\nFrozen Set")
print(fs)
```

```python
##Adding elements to Python Sets

people = {"Jay", "Idrish", "Archi"}

print("People:", end = " ")
print(people)
people.add("charan")
for i in range(1,6):
    people.add(i)
print("\nSet after adding element:", end = " ")
print(people)
```

```python
#Finding Differences of Sets in Python
set1 = set()
set2 = set()
for i in range(5):
    set1.add(i)

for i in range(3,9):
    set2.add(i)

set3 = set1.difference(set2)
print(" Difference of two sets using difference() function")
print(set3)

set3 = set1 - set2
print("\nDifference of two sets using '-' operator")
print(set3)
```
