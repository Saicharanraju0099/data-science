### Standard Library Overview

**The Python Standard Library is a comprehensive collection of pre-written modules and packages that come bundled with every Python installation. It provides a wide range of functionalities and tools, making it unnecessary for developers to write code from scratch for common programming tasks.**

```python
import array
arr = array.array('i',[1,2,3,4])
print(arr)
```

```python
## Data Serialization
import json
data = {"Name":"Charan","age":20}
json_str=json.dumps(data)
print(json_str)
print(type(json_str))

# converting back to dictionary
past_data=json.loads(json_str)
print(past_data)
print(type(past_data))
```

```python
##CSV (Comma Separated Values)
import csv
with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age',])
    writer.writerow(['Krish',25])

with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## Date and Time
from datetime import datetime,timedelta
now = datetime.now()
print(now)

yesterday = now - timedelta(days=1)
print(yesterday)
```

```python
# Time
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
# Regular Expression
import re
pattern = r"\d+"
text = "There are 123 Apples"
match = re.search(pattern,text)
print(match.group())
```
