**In Python, modules are single files containing Python code, while packages are directories containing multiple modules and a special __init__.py file (which can be empty). Importing allows you to use code defined in these modules or packages within other Python scripts.**

**Syntax:**
**from package_name import module_name**

```python
import math 
math.sqrt(6)
```

```python
from math import sqrt,pi
print(sqrt(9))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4,5])
```

```python
from math import *
print(sqrt(4))
print(pi)
```
